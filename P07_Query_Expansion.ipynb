# Query Expansion

`Query Expansion` adalah sebuah teknik dalam sistem temu kembali informasi untuk meningkatkan kinerja pencarian dengan memodifikasi kueri asli pengguna.

Intinya, teknik ini bertujuan untuk memperluas cakupan kueri dengan menambahkan istilah-istilah baru yang terkait, bersinonim, atau memiliki hubungan semantik dengan kata-kata dalam kueri awal.

## Tujuannya
Tujuan utama dari ekspansi kueri adalah untuk mengatasi kesenjangan kosakata `vocabulary mismatch` antara kueri pengguna dan dokumen yang relevan. Seringkali, dokumen yang relevan tidak menggunakan kata-kata yang sama persis dengan yang dimasukkan pengguna.

## Metode Umum
Ekspansi kueri dapat dilakukan melalui beberapa cara:
1. Menggunakan Tesaurus atau Kamus Sinonim: Mengganti atau menambahkan sinonim dari istilah kueri (misalnya, jika pengguna mencari "mobil", sistem menambahkan "kendaraan" atau "otomotif").

2. Analisis Umpan Balik Relevansi (Relevance Feedback):
    - Ekspansi Lokal (Pseudo-Relevance Feedback): Sistem menganggap beberapa dokumen peringkat teratas dari hasil pencarian awal sebagai dokumen yang relevan. Istilah-istilah penting dari dokumen-dokumen ini kemudian diekstrak dan ditambahkan ke kueri asli.

    - Ekspansi Global: Menggunakan informasi relevansi yang sudah dikumpulkan sebelumnya dari sesi pencarian masa lalu atau basis pengetahuan.

3. Menggunakan Korpus atau Model Bahasa: Memanfaatkan model bahasa (seperti Word Embeddings atau Large Language Models) untuk mengidentifikasi kata-kata yang secara semantik dekat dengan istilah kueri.

4. Koreksi Ejaan atau Normalisasi: Secara otomatis mengoreksi kesalahan ketik atau menggunakan bentuk akar kata (stemming/lemmatization) untuk mencocokkan lebih banyak variasi.

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.corpus import wordnet

def get_synonyms(term):

  synonyms = set()

  # cari sinonim dalam WordNet
  for syn in wordnet.synsets(term):
    for lemma in syn.lemmas():
      synonyms.add(lemma.name())
  return list(synonyms)

def query_expansion(query):
  expanded_terms = []

  query_terms = query.split()

  for term in query_terms:
    synonyms = get_synonyms(term)
    if synonyms:
      expanded_terms.extend(synonyms)
    else:
      expanded_terms.append(term)

  expanded_query = ' '.join(expanded_terms)
  return expanded_query

query = input("Masukkan Query asli: ")
expanded_query = query_expansion(query)
print("Query setelah di-expand: ", expanded_query)
```
