**Information Retrieval (IR)** adalah cabang dari ilmu komputer yang berfokus pada proses pengambilan informasi yang relevan dari koleksi data besar (seperti dokumen, teks, atau multimedia) berdasarkan permintaan pengguna. Secara sederhana, IR adalah "mesin" di balik pencarian informasi, seperti yang kita lihat di *Google Search*, sistem rekomendasi di *Netflix*, atau pencarian di perpustakaan digital.
- Contoh Aplikasi: Saat kamu mencari `Sahroni` di mesin pencari, IR bekerja untuk menemukan dokumen-dokumen yang paling cocok dari miliaran halaman web.
- Komponen Utama: Melibatkan indexing (penyusunan data), query processing (pemrosesan pertanyaan pengguna), dan ranking (pengurutan hasil berdasarkan relevansi).
- Perbedaan dengan Database: IR lebih fleksibel untuk data tidak terstruktur (seperti teks bebas), sementara database lebih untuk data terstruktur seperti SQL.
